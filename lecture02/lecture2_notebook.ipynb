### 3-2　目的変数を比較変数で集計する

③目的変数を④比較変数で分けて平均を見ます。

```python
# 3-2　目的変数を比較変数で集計する
# TODO: ③目的変数と④比較変数を入れてください
# 例: df.groupby('workingday')['cnt'].mean()

result = df.groupby("hr")["cnt"].mean()
print(result.round(1))
```

### 3-3【発展】可視化する

時間に余裕があれば、自分の問いに合ったグラフを作成してください。

```python
# 3-3　可視化する
# TODO: 自分の問いに合ったグラフを選んで作成してください
#
# 選択肢：
#   棒グラフ  → ax.bar(result.index, result.values)
#   折れ線    → ax.plot(result.index, result.values, marker='o')
#   箱ひげ図  → df.boxplot(column=____, by=____, ax=ax)

weekday_avg = df[df['workingday'] == 1].groupby('hr')['cnt'].mean()
holiday_avg = df[df['workingday'] == 0].groupby('hr')['cnt'].mean()

# （ここにグラフのコードを書く）

fig, ax = plt.subplots()
ax.plot(weekday_avg.index, weekday_avg.values, label='Weekday', color='#1E2761',
        lw=2.5, marker='o', markersize=4)
ax.plot(holiday_avg.index, holiday_avg.values, label='Holiday', color='#C0392B',
        lw=2.5, marker='o', markersize=4)
ax.set_xlabel('Hour')
ax.set_ylabel('Avg Count')
ax.set_title('Peak time : Weekday vs Holiday')
ax.set_xticks(range(0, 24, 1))
ax.legend(fontsize=12)
plt.tight_layout()
plt.show()
```
